# Template: Sandbox Notebook

This notebook serves as a **template** for creating sandbox analysis notebooks.

**Key Requirements for Sandbox Notebooks:**
1.  **No Hardcoded Paths**: Always use `ProjectPaths` from `utils.paths`.
2.  **Outputs**: Save any results (data or figures) to `figures/step0_sandbox/<script_name>` as per the global rules. Do not clutter the root sandbox directory.

```
# --- 1. Imports ---
import pandas as pd
import matplotlib.pyplot as plt
import sys
from pathlib import Path
import os

# Ensure we can import from 'utils' directory
# This allows running the notebook from anywhere
project_root = Path("../../").resolve()
if str(project_root) not in sys.path:
    sys.path.append(str(project_root))

from utils.paths import ProjectPaths
```

## 2. Output Directory Setup

Configure the output directory for this specific sandbox script.

```
# Define the name of this notebook for output folder creation
notebook_name = "101_example_notebook" 

# Setup output directory specific to this sandbox notebook
# According to global rules, sandbox outputs go to figures/step0_sandbox/<script_name>
sandbox_output_dir = ProjectPaths.FIGURES_DIR / "step0_sandbox" / notebook_name
sandbox_output_dir.mkdir(parents=True, exist_ok=True)
print(f"Output directory set to: {sandbox_output_dir}")
```

## 3. Data Loading

Use `ProjectPaths` to load inputs.

```
# Example: Loading some data
# input_path = ProjectPaths.PROCESSED_DATA_DIR / "some_data.csv"
# df = pd.read_csv(input_path)

# Dummy data for template demonstration
df = pd.DataFrame({'X': [1, 2, 3], 'Y': [4, 5, 6]})
display(df.head())
```

## 4. Analysis & Visualization

Perform analysis and save outputs to the designated sandbox output directory.

```
# Perform analysis and plot
plt.figure(figsize=(6, 4))
plt.scatter(df['X'], df['Y'], color='blue', alpha=0.6)
plt.title("Sandbox Example Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)

# Save the figure to the sandbox output directory
fig_out = sandbox_output_dir / "example_sandbox_plot.png"
plt.savefig(fig_out)
print(f"Saved figure to: {fig_out}")

plt.show()
```
