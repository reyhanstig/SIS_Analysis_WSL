```python
import numpy as np
import matplotlib.pyplot as plt
```

Gaussian curve. Use this later to fit with the data correlation plotting.
This will later be used to plan the observation plan. Use decorrelation scale as a reference to determine how far can two sampling locations can be before they are expected to have the same value or not.

```python
dx = 20
x = np.linspace(-50, 50, 400)
e = np.exp(-(x/dx)**2) - 0.2 # <-- 0.2 offset set the floor of the curve lower

plt.figure()
plt.plot(x, e)
```

```python
def gaussian_curve(h, a, c):
    return c * (1 - np.exp(-(h**2) / (a**2)))

plt.figure()
plt.plot(x, gaussian_curve(x, dx, 1))
```
