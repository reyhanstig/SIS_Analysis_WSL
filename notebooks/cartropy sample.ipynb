```python
# Plotting countries from Natural Earth shapefile using Cartopy
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.io.shapereader as shpreader

# --- 1. Set up the map ---
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(1, 1, 1, projection=ccrs.PlateCarree())

# --- 2. Get the shapefile path ---
# Cartopy has a helper to get the path to its built-in datasets
shpfilename = shpreader.natural_earth(resolution='110m',
                                      category='cultural',
                                      name='admin_0_countries')

# --- 3. Read the geometries from the shapefile ---
# We use .geometries() because we only need the shapes for plotting
geometries = shpreader.Reader(shpfilename).geometries()

# --- 4. Add the geometries to the map axes ---
ax.add_geometries(geometries, 
                  crs=ccrs.PlateCarree(), # The CRS of the shapefile data
                  edgecolor='black', 
                  facecolor='lightgray')

# --- 5. Finalize and show the plot ---
ax.coastlines()
ax.set_title("Countries from Natural Earth Shapefile")
plt.show()
```
