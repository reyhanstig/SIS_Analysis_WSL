# Testing Global `matplotlibrc` Configuration
This notebook tests the global parameters defined in `matplotlibrc` at the root of the project.

```python
# Load global style file
import matplotlib.pyplot as plt
plt.style.use("../project_style.mplstyle")

print("Project style loaded.")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Create dummy data resembling the PCA scatter plots
np.random.seed(42)
n_points = 50
x = np.random.randn(n_points) * 0.2
y = np.random.randn(n_points) * 0.1

divisions = ['Osaka Bay', 'Kii Channel', 'Harima Nada', 'Bisan Seto', 'Hiroshima Bay']
colors = ['tab:blue', 'tab:orange', 'tab:green', 'tab:red', 'tab:purple']

fig, ax = plt.subplots(figsize=(10, 6))

for i, div in enumerate(divisions):
    # Subset points for this division
    idx = range(i, n_points, len(divisions))
    ax.scatter(x[idx], y[idx], label=div, c=colors[i], edgecolors='k', alpha=0.7)
    
    # Add dummy text annotations (similar to the station ID numbers in the images)
    for j in list(idx)[:3]:  
        ax.text(x[j]+0.005, y[j]+0.005, f"000{j+1000}", 
        # fontsize=10 # manually define font size for the scatter plot points.
        ) 
ax.axhline(0, color='gray', linestyle='--', linewidth=1)
ax.axvline(0, color='gray', linestyle='--', linewidth=1)

# These parameters should be automatically styled by the matplotlibrc file
ax.set_title('DO (PC1 Loading) vs. Spring Chl-a Correlation')
ax.set_xlabel('Spring Chl-a (Pearson r with PC1)')
ax.set_ylabel('DO (PC1 Loading)')

# Add legend outside the plot
ax.legend(title='Division', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()
```
